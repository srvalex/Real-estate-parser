```python
import json
import requests

PUBLIC_URL = "https://josue-unbiddable-unreproachably.ngrok-free.dev"   # paste your URL here
ENDPOINT   = f"{PUBLIC_URL}/extract-keywords"

user_prompt = (
    "Apartament pet friendly cu 2 camere la 10 minute de metrou"
)

response = requests.post(ENDPOINT, json={"text": user_prompt})

if response.status_code == 200:
    print("✅  Server returned:")
    print(json.dumps(response.json(), indent=2, ensure_ascii=False))
else:
    print(f"❌  Error {response.status_code}: {response.text}")
```
